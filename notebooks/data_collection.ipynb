# Data Reorganising

## Nazi symbol detection
Organise the downloaded datasets for nazi symbol binary classification and split them into new train, val and test datasets.

```python
import math

def split_images(images, train_size=None, val_size=None, test_size=None):
    train_size = train_size or 0.7
    val_size = val_size or 0.15
    test_size = test_size or 0.15
    if train_size + val_size + test_size != 1:
        raise ValueError
    train, val, test = [], [], []
    label_dict = dict()
    for image in images:
        label = image.split("/")[-2]
        if label not in label_dict:
            label_dict[label] = []
        label_dict[label].append(image)
    for k, v in label_dict.items():
        train_boundary = math.floor(len(v) * train_size)
        train += v[:train_boundary]
        val_boundary = math.floor(len(v) * (train_size + val_size))
        val += v[train_boundary:val_boundary]
        test += v[val_boundary:]
    return train, val, test
```

```python
train, val, test = [], [], []
```

```python
for images in [sports_images, car_bike_images, people_clothing_images, fruit_vegetable_images, animal_images, 
               kids_and_adults_images, scooby_doo_images, low_light_animal_images, ai_humen_generated_images]:
    tmp_train, tmp_val, tmp_test = split_images(images)
    train += tmp_train
    val += tmp_val
    test += tmp_test
```

```python
for image in train:
    shutil.copy(image, "/home/zhiwei/Projects/nazi-symbols-classification/datasets/nazi-symbols-detection/train/non-nazi/")

for image in val:
    shutil.copy(image, "/home/zhiwei/Projects/nazi-symbols-classification/datasets/nazi-symbols-detection/val/non-nazi/")

for image in test:
    shutil.copy(image, "/home/zhiwei/Projects/nazi-symbols-classification/datasets/nazi-symbols-detection/test/non-nazi/")
```

```python
import os
import shutil

def reorganise_images(path="/home/zhiwei/Projects/nazi-symbols-classification/datasets/nazi-symbols-classification"):
    sub_folders = ["train"]
    for sub_folder_name in sub_folders:
        sub_folder_path = os.path.join(path, sub_folder_name)
        for dir_name in os.listdir(sub_folder_path):
            if dir_name.startswith("."):
                continue
            elif " " in dir_name:
                classes = dir_name.split(" ")
                dir_path = os.path.join(sub_folder_path, dir_name)
                for class_dir in classes:
                    shutil.copytree(dir_path, os.path.join(sub_folder_path, class_dir), dirs_exist_ok=True)
                shutil.rmtree(dir_path)
```
