## Organise the downloaded datasets and split them into new train, val and test datasets.

```python
car_bike_images = get_image_paths("/Users/zhiweizhang/Projects/nazi_symbols_classification/notebooks/datasets/Car-Bike-Dataset", sub_folders=None)
```

```python
fruit_vegetable_images = get_image_paths("/Users/zhiweizhang/Projects/nazi_symbols_classification/notebooks/datasets/fruit-and-vegetable-image-recognition", sub_folders=("train", "test", "validation"))
```

```python
people_clothing_images = get_image_paths("/Users/zhiweizhang/Projects/nazi_symbols_classification/notebooks/datasets/people-clothing-segmentation", sub_folders=("jpeg_images", "png_images"))
```

```python
sports_images = get_image_paths("/Users/zhiweizhang/Projects/nazi_symbols_classification/notebooks/datasets/sports-classification", sub_folders=("train", "test", "valid"))
```

```python
sports_images = [image for image in sports_images if image.endswith("png") or image.endswith("jpg") or image.endswith("jpeg")]
```

```python
def split_images(images):
    train = []
    val = []
    test = []
    label_dict = dict()
    for image in images:
        label = image.split("/")[-2]
        if label not in label_dict:
            label_dict[label] = []
        label_dict[label].append(image)
    for k, v in label_dict.items():
        boundary = len(v)//4
        train += v[:boundary]
        val += v[boundary:boundary*2]
        test += v[boundary*2:]
    return train, val, test
```

```python
train, val, test = [], [], []
```

```python
for images in [animal_images, car_bike_images, fruit_vegetable_images, people_clothing_images, sports_images]:
    tmp_train, tmp_val, tmp_test = split_images(images)
    train += tmp_train
    val += tmp_val
    test += tmp_test
```
