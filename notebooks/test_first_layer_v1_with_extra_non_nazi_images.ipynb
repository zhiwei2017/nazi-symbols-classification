```python
from nazi_symbols_classification_backend.app.services.classification import get_first_layer_result, init_state_for_classification
```

```python
init_state_for_classification()
```

```python
from sklearn.metrics import classification_report
```

```python
def evaluate(images):
    result = []
    for i in range(0, len(images), 100):
        result += get_first_layer_result(images[i:i+100])
    result = [d['first_layer_result']['label'] for d in result]
    y_true = ["non-nazi"] * len(result)
    print(classification_report(y_true, result))
    return result
```

```python
evaluate(test_images)
```
